```python
import numpy as np
```

```python
df=pd.read_csv("data.csv")
print(df)
```

```python
# maxsalary=df["Name"].where([df["Salary"]].max())


# Find the index of the maximum salary
max_salary_index = df['Salary'].idxmax()
print(max_salary_index)

# Use the index to get the corresponding name
max_salary_name = df["Name"].iloc[max_salary_index]
print(max_salary_name)

# Sort the DataFrame by 'Salary' in descending order and select the first row
max_salary_row = df.sort_values(by='Salary', ascending=False).iloc[0]

print(max_salary_row)

```

```python
min_salary_row = df.sort_values(by='Salary').iloc[0]
print(min_salary_row )
```

```python
df["Age"] = df["Age"].sort_values(ascending = False)
print(df["Age"])

Age= df.sort_values(by = 'Age',ascending = False)
print(Age)
```

```python

# 1. Check the shape
print("Initial DataFrame shape:", df.shape)

# 2. Add the new row
new_row_data = {'Name': 'Parita', 'Age': 32, 'Salary': 25000}
df.loc[len(df)] = new_row_data

# Check the updated DataFrame and its new shape
print("\nDataFrame after adding the new row:")
print(df)
print("\nUpdated DataFrame shape:", df.shape)
```
